 1.  Category Performance Analysis

-  Objective: Compare total Sales by Product Category.

Steps in Tableau:

Connect the Retail_Dataset to Tableau.
Open a new worksheet.
Drag Product Category to the Columns shelf.
Drag Sales to the Rows shelf.
Ensure the aggregation is SUM(Sales).
Select Bar Chart.
Sort the bars in descending order.

Insight:
The category with the tallest bar has the highest total sales and is therefore the top-performing category in terms of revenue.

 2.  Sales Trend Analysis

-  Objective: Analyze monthly sales trends.

Steps:

Open a new worksheet.
Drag Order Date to the Columns shelf.
Change the date level to Month.
Drag Sales to the Rows shelf.
Select Line Chart.
Use SUM(Sales) as the aggregation.

Insight:
The line chart helps identify peak and slow sales periods. Management can use peak periods for promotions and ensure sufficient inventory, while slow periods can be targeted with discounts or marketing campaigns.

3.  Regional Contribution Analysis

-  Objective: Compare sales generated by different regions.

Steps:

Open a new worksheet.
Drag Region to the Columns shelf.
Drag Sales to the Rows shelf.
Use SUM(Sales).
Select a Bar Chart.
Sort the regions from highest to lowest sales.

Insight:
The region with the highest bar is the highest contributing region.

A Pie Chart can also be used to show each region's percentage contribution to total sales, but a bar chart is generally easier for accurate comparison.

4.  Profitability Investigation

-  Objective: Identify high-sales but low-profit or loss-making orders.

Steps:

Open a new worksheet.
Drag Sales to the Columns shelf.
Drag Profit to the Rows shelf.
Select Scatter Plot.
Add Order ID or Product to Detail on the Marks card.
Add Profit to Color.
Identify records where Profit is below zero.

Insight:
Orders with high Sales but negative or very low Profit indicate potential profitability problems. These may be caused by high discounts, product costs, shipping costs, or pricing issues.

The area where Profit < 0 represents loss-making orders.

 5. Business Growth Evaluation

-  Objective: Visualize cumulative Sales growth over time.

Steps:

Drag Order Date to the Columns shelf.
Select Month or Year-Month.
Drag Sales to the Rows shelf.
Right-click SUM(Sales).
Select Quick Table Calculation → Running Total.
Select Line Chart.

This produces a cumulative/running sales line.

Insight:
A continuously increasing cumulative sales line indicates positive overall sales growth. Flat periods indicate slower growth, while sharp increases show periods of strong sales.

6. Category vs Region Performance

-  Objective: Compare sales of different categories across regions.

Steps:

Open a new worksheet.
Drag Region to the Columns shelf.
Drag Product Category to the Rows shelf.
Drag Sales to the Marks → Color or use it as a measure.
Select Bar Chart or Heat Map.

A grouped bar chart can also be created by placing:

Region → Columns
Product Category → Columns
Sales → Rows

Insight:
This analysis identifies which category performs strongly in each region and which category-region combinations need improvement.

7.  Product Prioritization

-  Objective: Identify products that create maximum business impact using Sales, Profit, and Quantity.

Steps:

Open a new worksheet.
Drag Product to Detail.
Drag Sales to Columns.
Drag Profit to Rows.
Drag Quantity to Size.
Use Profit or Sales on Color.
Create a Scatter Plot.
Interpretation

Products can be classified as:

High Sales + High Profit + High Quantity: Priority products.
High Sales + Low/Negative Profit: Products requiring profitability investigation.
Low Sales + High Profit: Potential products for marketing and growth.
Low Sales + Low Profit: Products that may need review or replacement.

This gives management a better understanding of product performance than looking at Sales alone.

 8.  Executive Dashboard Creation

-  Objective: Create a single interactive dashboard for the CEO.

Recommended visualizations:
Sales by Category – Bar Chart
Monthly Sales Trend – Line Chart
Regional Sales – Bar Chart
Sales vs Profit – Scatter Plot
Dashboard steps:
Create the required worksheets.
Click New Dashboard.
Drag the worksheets onto the dashboard.
Arrange them clearly.
Add a Region filter.
Add a Product Category filter.
Apply the filters to the relevant worksheets.
Add suitable titles.
Add important KPIs such as:
Total Sales
Total Profit
Total Quantity
Format the dashboard for easy reading.

Result:
The CEO can select a region or category and immediately see how sales, profit, and trends change.


9. Chart Selection Justification

Different charts are used because different business questions require different types of analysis.

-  | Chart                        | Purpose                                                       |
| ---------------------------- | ------------------------------------------------------------- |
| **Bar Chart**                | Best for comparing Sales between categories or regions.       |
| **Line Chart**               | Best for identifying Sales trends and changes over time.      |
| **Scatter Plot**             | Best for analyzing the relationship between Sales and Profit. |
| **Heat Map**                 | Useful for comparing Category performance across Regions.     |
| **Running Total Line Chart** | Useful for evaluating cumulative Sales growth.                |



10. Business Insight Summary
-   Insight 1: Category and Regional Performance

The dashboard can identify the highest-selling product category and the region contributing the most sales. Management can focus inventory, marketing, and sales resources on successful category-region combinations while improving weaker markets.

Insight 2: Sales vs Profitability

The Sales vs Profit analysis can reveal high-sales products or orders that generate very low or negative profit. Management can investigate pricing, discounts, and costs for these products to improve overall profitability.

How these insights help management

These insights help management to:

Improve sales strategies.
Focus on high-performing products and regions.
Reduce losses from low-profit orders.
Plan inventory more effectively.
Design targeted promotions.
Make data-driven decisions for the next quarter.
